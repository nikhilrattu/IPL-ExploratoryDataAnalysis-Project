# Loading Libraries and Dataset

```python
import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

```

### Load the files

```python
df = pd.read_csv('IPL.csv')
df.head()
```

#### Basic Info

```python
df.info()
```

#### check null values

```python
df.isnull().sum()
```

##### Which team won the most number of matches?

```python
most_match_wins=df['match_winner'].value_counts()
sns.barplot(y=most_match_wins.index,x=most_match_wins.values,palette='Set2',hue=most_match_wins.index)
plt.title('Most Matches Won by Team')
```

##### Toss Decision Trends
###### means whether teams decide to bat or bowl first.

```python
toss_des=df['toss_decision'].value_counts()
sns.countplot(x=df['toss_decision'],hue=df['toss_decision'])
plt.title('Toss Decision trend')
```

##### Toss winner vs match winner

```python
decs= df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage = (100 * decs/df.shape[0]).round(2)
percentage
```

```python
px.pie(names=["Won Toss & Match", "Lost After Winning Toss"],values=[percentage,100-percentage],hole=0.5)
```

##### How did teams win?(Runs or Wickets)

```python
df['won_by'].value_counts()
sns.countplot(df,x='won_by',palette='Set2')
```

#### **Key Performances**

##### Most player of the match awards

```python
player_award=df['player_of_the_match'].value_counts().head(5)
sns.barplot(y=player_award.index,x=player_award.values,palette='Set2')
plt.title('Player Of The Match')
```

Top 2 Scorers 

```python
df.head()
```

```python
top_score=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
top_score
px.bar(top_score)
```

```python
df['most_wickets']=df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['most_wickets']=df['most_wickets'].astype(int)
df
top_bowlers=df.groupby('best_bowling')['most_wickets'].sum().sort_values(ascending=False).head(5)
top_bowlers=top_bowlers.reset_index()
top_bowlers.columns=['players','wickets']
top_bowlers
px.bar(top_bowlers,y='players',x='wickets')
```

##### Most Matches played by venue
